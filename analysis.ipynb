Проверяем число поездок и сравниваем час утреннего пика для пользователей разных типов. Перед запуском выполните `python pipeline.py`.

```python
import pandas as pd
hourly = pd.read_csv('site/hourly.csv')
work = hourly.query('working == 1 and group != "cnt" and 5 <= hour <= 11')
work.loc[work.groupby(['year','group'])['mean'].idxmax(), ['year','group','hour','mean','days','lo','hi']]
```

```python
import seaborn as sns
sns.lineplot(data=work, x='hour', y='mean', hue='group', style='year', markers=True, errorbar=None)
```
